**DICTIONARIES**

```python
#Creating dictionaries
d = {'rno':503,'Name':'Ramteja','Life':'Single'}
print(d)
print(d['Name'])
d['Name']='Ram'
print(d)
```

```python
students = {'stud1':['Madhu',18],'stud2':['Ram',19],'stud3':['Abhi',20]}
print(students)
print(students['stud2'])
print(students.keys())
print(students.values())
```

**BUILT IN FUNCTIONS**

```python
a = {'Name':'Ram','Marks':35,'Life':'Single'}
print(len(a))
print(sorted(a))
print(max(a))
print(min(a))
```

```python
#Write a program to check if a given key exists in a dictionary or not
a = {'Name':'Ram','rno':4503}
b = input()
if b in a:
  print('Yes')
else:
  print('No')
```

```python
#Write a program to add new key-value  pair to an existing dictionary
a = {'Name':'Ram','rno':4503}
a['grade'] = 8
print(a)
a.update({'Height':6,'village':'Amalapuram'})
print(a)
```

```python
a = {1:10,2:20,3:30}
print(sum(a.values()))
```

```python
#Write a program that prints frequency of each element in a list in following
#format [1,2,3,4,1,5,6,1,6,7]     {1:3,2:1,3:1,4:1,5:1,6:2,7:1}
a = [int(i) for i in input().split()]
d = {}
for i in a:
  d[i] = a.count(i)
print(d)
```
